**Q11.** Given student marks, calculate class total and average.

```python
# Write your solution here
marks = np.array([75, 80, 65, 90, 70])

class_total = np.sum(marks)
class_average = np.mean(marks)

print("Class Total:", class_total)
print("Class Average:", class_average)
```

**Q12.** Given employee salaries, calculate average, minimum and maximum.

```python
# Write your solution here
salaries = np.array([25000, 30000, 45000, 50000, 35000])

average = np.mean(salaries)
minimum = np.min(salaries)
maximum = np.max(salaries)

print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

**Q13.** Given daily sales, find the best and worst sales day.

```python
# Write your solution here
sales = np.array([1200, 1500, 900, 1800, 1100, 2000, 1300])

best_day = np.argmax(sales)
worst_day = np.argmin(sales)

print("Best day:", best_day + 1)
print("Best sales:", sales[best_day])

print("Worst day:", worst_day + 1)
print("Worst sales:", sales[worst_day])
```
